```python
import numpy as np

class RegressionLineaire:
    def __init__(self, add_b=False, b=None, x=None, xtrain=None, ytrain=None):
        self.train(xtrain, ytrain)
        if add_b:
            x = np.insert(x, 0, 1, axis=1)
            self.w = np.insert(self.w, 0, b)
        self.ychapeau = self.predict(x)

    def predict(self, x):
        return x @ self.w

    def train(self, x, y):
        self.w = np.linalg.inv(x.T @ x) @ (x.T @ y)


#xtrain and ytrain from regression examples of 'Overview of Machine Learning' (house prices)
xtrain = np.array([[7420, 4, 2, 3, 1, 0, 0, 1],
                   [8968, 4, 4, 4, 1, 0, 0, 1],
                   [9960, 3, 2, 2, 1, 1, 0, 0],
                   [7500, 4, 2, 2, 1, 1, 1, 3],
                   [7420, 4, 1, 2, 1, 1, 1, 2],
                   [7500, 3, 3, 1, 1, 0, 1, 2],
                   [8500, 4, 3, 4, 1, 0, 0, 1],
                   [16200, 5, 3, 2, 1, 0, 0, 0],
                   [8100, 4, 1, 2, 1, 1,  1, 2],
                   [5750, 3, 2, 4, 1, 1, 1, 1],
                   [13200, 3, 1, 2, 1, 0, 1, 2]])

ytrain = np.array([13300, 12250, 12250, 12215, 11410, 10850, 10150, 10150, 9870, 9800, 9880])

#x=xtrain for test

#with b=13 (i take it randomly)
rlb=RegressionLineaire(add_b=True,b=13,x=xtrain,xtrain=xtrain,ytrain=ytrain)
print(rlb.ychapeau)

#without b
rl=RegressionLineaire(x=xtrain,xtrain=xtrain,ytrain=ytrain)
print(rl.ychapeau)


```
